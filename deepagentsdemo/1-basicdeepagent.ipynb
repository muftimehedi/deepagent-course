### Deep Agent Overview
Deep Agents are an open-source "batteries-included" agent harness designed by LangChain to handle complex, long-running, and multi-step tasks like coding and deep research.
Unlike traditional or "shallow" agents that operate entirely within a single, ephemeral LLM context window and lose track of goals over hundreds of steps, Deep Agents externalize state, planning, and memory into robust underlying structures. You can explore the project via the Deep Agents Documentation or read more on the LangChain Deep Agents Page or LangChain Deep Agents Overview.
Core Capabilities & Architecture
• Virtual Filesystem & Context Management: Uses a pluggable virtual file system to offload large tool results and conversation history past token limits, keeping the core context clean via progressive disclosure.
• Task Planning: Employs explicit planning and todo-list tools so the main coordinator agent can decompose objectives, track progress, and adapt dynamically.
• Subagent Delegation: Spawns specialized, parallel subagents running in isolated context windows, preventing procedural noise from polluting the main agent's working memory.
• Long-Term Memory & Skills: Loads persistent preferences and instructions from AGENTS.md files and on-demand domain knowledge from SKILL.md files.
• Safety & Control: Features built-in sandboxed code execution, human-in-the-loop approval gates for critical decisions, and declarative filesystem access permissions.
How It Fits in the Stack
Deep Agents sit as a meta-toolkit one level above regular AI frameworks, built on top of LangChain's core abstractions and powered by the LangGraph runtime for durable execution:
• LangChain: Provides generic building blocks for chat models and tools.
• LangGraph: Delivers low-level agent runtime infrastructure like checkpointing, streaming, and state management.
• Deep Agents: Acts as an opinionated harness pre-configured with middleware for planning, file systems, and multi-agent coordination out of the box.

### When to use deep agent?
Use a deep agent when you are building autonomous systems for complex, long-running, multi-step tasks that require explicit task decomposition, context offloading, and subagent delegation over extended time horizons
